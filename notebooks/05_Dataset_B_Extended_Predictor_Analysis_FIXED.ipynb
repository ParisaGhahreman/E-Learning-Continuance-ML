# Dataset B Extended Predictor Analysis

This notebook evaluates whether Dataset B-specific constructs provide additional predictive value for students' e-learning continuance intention.

The analysis extends the cross-dataset core model by examining:

- Perceived Usefulness (PU)
- Perceived Educational Support (PEdS)
- Perceived Emotional Support (PEmS)

The outcome variable is Continuance Intention (CI).

The core cross-dataset model remains:

CON + SAT → CI

The additional constructs are evaluated only within Dataset B because equivalent PEdS and PEmS constructs are not available in the external EFL dataset.

The purpose of this analysis is to assess incremental predictive value, not causal effects.

```python
# در این سلول کتابخانه‌های موردنیاز و مسیرهای Notebook 5 به‌صورت خودکار تعریف می‌شوند.

from pathlib import Path
import os
import pandas as pd
import numpy as np


# مسیر فعلی اجرای Notebook
CURRENT_DIR = Path.cwd().resolve()


# شناسایی خودکار ریشه پروژه
if CURRENT_DIR.name == "notebooks":

    PROJECT_ROOT = CURRENT_DIR.parent

elif (
    (CURRENT_DIR / "notebooks").exists()
    and (CURRENT_DIR / "data").exists()
):

    PROJECT_ROOT = CURRENT_DIR

else:

    raise FileNotFoundError(
        "Project root could not be detected. "
        "Please run this notebook from the project folder or the notebooks folder."
    )


PROCESSED_FOLDER = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed"
)

TABLES_FOLDER = os.path.join(
    PROJECT_ROOT,
    "outputs",
    "tables"
)


B_DATA_PATH = os.path.join(
    PROCESSED_FOLDER,
    "dataset_B_processed.csv"
)


print("Project root:")
print(PROJECT_ROOT)

print("\nDataset B processed file:")
print(B_DATA_PATH)

print("\nTables folder:")
print(TABLES_FOLDER)


print("\nPath check:")
print(
    "Dataset B exists:",
    os.path.exists(B_DATA_PATH)
)

print(
    "Tables folder exists:",
    os.path.exists(TABLES_FOLDER)
)
```

```python
# در این سلول نسخه پردازش‌شده Dataset B برای تحلیل Predictorهای توسعه‌یافته بارگذاری می‌شود.

df_B_extended = pd.read_csv(
    B_DATA_PATH
)

print("Dataset B shape:")
print(df_B_extended.shape)

print("\nDataset B columns:")

for column in df_B_extended.columns:
    print(column)
```

```python
# در این سلول وجود Compositeهای موردنیاز، مقادیر گمشده و آمار اولیه آن‌ها بررسی می‌شود.

extended_columns = [
    "PEdS_score",
    "PEmS_score",
    "PU_score",
    "CON_score",
    "SAT_score",
    "CI_score"
]


print("Required columns available:")

for column in extended_columns:

    print(
        column,
        ":",
        column in df_B_extended.columns
    )


print("\nMissing values:")

print(
    df_B_extended[
        extended_columns
    ].isna().sum()
)


print("\nConstruct statistics:")

display(
    df_B_extended[
        extended_columns
    ].describe().round(3)
)
```

```python
# در این سلول پنج مجموعه Predictor برای بررسی ارزش افزوده سازه‌های اختصاصی Dataset B تعریف می‌شوند.

predictor_sets = {

    "M1_Core": [
        "CON_score",
        "SAT_score"
    ],

    "M2_Core_PU": [
        "PU_score",
        "CON_score",
        "SAT_score"
    ],

    "M3_Core_PU_PEdS": [
        "PU_score",
        "PEdS_score",
        "CON_score",
        "SAT_score"
    ],

    "M4_Core_PU_PEmS": [
        "PU_score",
        "PEmS_score",
        "CON_score",
        "SAT_score"
    ],

    "M5_Full": [
        "PU_score",
        "PEdS_score",
        "PEmS_score",
        "CON_score",
        "SAT_score"
    ]
}

for model_name, predictors in predictor_sets.items():
    print(model_name, "->", predictors)
```

```python
# در این سلول تقسیم داده، Repeated Cross-Validation و پنج الگوریتم ثابت برای مقایسه Predictorها تعریف می‌شوند.

from sklearn.model_selection import (
    train_test_split,
    RepeatedKFold,
    cross_validate
)

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)


all_indices = np.arange(
    len(df_B_extended)
)

train_indices, test_indices = train_test_split(
    all_indices,
    test_size=0.20,
    random_state=42
)

y_extended = df_B_extended[
    "CI_score"
]


repeated_cv = RepeatedKFold(
    n_splits=5,
    n_repeats=10,
    random_state=42
)


scoring = {
    "MAE": "neg_mean_absolute_error",
    "MSE": "neg_mean_squared_error",
    "R2": "r2"
}


extended_models = {

    "Linear Regression":
        LinearRegression(),

    "KNN":
        Pipeline([
            (
                "scaler",
                StandardScaler()
            ),
            (
                "knn",
                KNeighborsRegressor(
                    n_neighbors=17
                )
            )
        ]),

    "Decision Tree":
        DecisionTreeRegressor(
            max_depth=4,
            min_samples_leaf=1,
            min_samples_split=10,
            random_state=42
        ),

    "Random Forest":
        RandomForestRegressor(
            max_depth=3,
            min_samples_leaf=1,
            min_samples_split=10,
            n_estimators=300,
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingRegressor(
            learning_rate=0.05,
            max_depth=2,
            min_samples_leaf=1,
            n_estimators=50,
            random_state=42
        )
}


print("Training samples:", len(train_indices))
print("Testing samples :", len(test_indices))
print("Repeated CV     : 5 folds × 10 repeats")

print("\nAlgorithms:")

for algorithm_name in extended_models:
    print(algorithm_name)
```

```python
# در این سلول پنج مجموعه Predictor روی پنج الگوریتم با Repeated Cross-Validation مقایسه می‌شوند.

all_model_results = []

for predictor_set_name, predictors in predictor_sets.items():

    X_model = df_B_extended[predictors]
    X_train_model = X_model.iloc[train_indices]
    y_train_model = y_extended.iloc[train_indices]

    for algorithm_name, algorithm in extended_models.items():

        scores = cross_validate(
            algorithm,
            X_train_model,
            y_train_model,
            cv=repeated_cv,
            scoring=scoring,
            n_jobs=-1
        )

        mae_scores = -scores["test_MAE"]
        rmse_scores = np.sqrt(-scores["test_MSE"])
        r2_scores = scores["test_R2"]

        all_model_results.append({
            "Predictor_Set": predictor_set_name,
            "Algorithm": algorithm_name,
            "Number_of_Predictors": len(predictors),
            "MAE_Mean": mae_scores.mean(),
            "MAE_Std": mae_scores.std(),
            "RMSE_Mean": rmse_scores.mean(),
            "RMSE_Std": rmse_scores.std(),
            "R2_Mean": r2_scores.mean(),
            "R2_Std": r2_scores.std()
        })

all_algorithms_comparison = pd.DataFrame(all_model_results)

metric_columns = [
    "MAE_Mean",
    "MAE_Std",
    "RMSE_Mean",
    "RMSE_Std",
    "R2_Mean",
    "R2_Std"
]

all_algorithms_comparison[metric_columns] = (
    all_algorithms_comparison[metric_columns].round(4)
)

print(
    "Number of evaluated combinations:",
    len(all_algorithms_comparison)
)

display(all_algorithms_comparison)
```

```python
# در این سلول RMSE تمام Predictor Setها در پنج الگوریتم به شکل یک جدول مقایسه‌ای خوانا ساخته می‌شود.

rmse_pivot = all_algorithms_comparison.pivot(
    index="Algorithm",
    columns="Predictor_Set",
    values="RMSE_Mean"
)

rmse_pivot = rmse_pivot[
    [
        "M1_Core",
        "M2_Core_PU",
        "M3_Core_PU_PEdS",
        "M4_Core_PU_PEmS",
        "M5_Full"
    ]
]

rmse_pivot = rmse_pivot.round(4)

display(rmse_pivot)
```

```python
# در این سلول دو جدول نهایی تحلیل Predictorهای Dataset B ذخیره می‌شوند.

all_results_path = os.path.join(
    TABLES_FOLDER,
    "Dataset_B_Predictor_Sets_All_Algorithms.xlsx"
)

rmse_table_path = os.path.join(
    TABLES_FOLDER,
    "Dataset_B_RMSE_Predictor_Set_Comparison.xlsx"
)

all_algorithms_comparison.to_excel(
    all_results_path,
    index=False
)

rmse_pivot.to_excel(rmse_table_path)

print("Saved:")
print(all_results_path)

print("\nSaved:")
print(rmse_table_path)
```
