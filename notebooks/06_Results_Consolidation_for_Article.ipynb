# Article Results Consolidation

این Notebook برای تجمیع نتایج نهایی پروژه و آماده‌سازی جداول مورد نیاز مقاله ایجاد شده است.

هیچ مدل جدیدی در این Notebook آموزش داده نمی‌شود و هیچ Hyperparameter یا تصمیم تحلیلی تغییر نمی‌کند.

تمام نتایج از خروجی‌های نهایی Notebookهای قبلی خوانده می‌شوند تا بین کد، جداول و متن مقاله سازگاری کامل حفظ شود.

## مراحل پروژه

1. Data Audit and Quality Control
2. Construct Harmonization
3. Model Development and Selection
4. External Validation and Sensitivity Analysis
5. Dataset B Extended Predictor Analysis
6. Article Results Consolidation

```python
# در این سلول کتابخانه‌های موردنیاز و مسیرهای اصلی پروژه به‌صورت خودکار تعریف می‌شوند.

from pathlib import Path
import os
import pandas as pd
import numpy as np


# مسیر فعلی اجرای Notebook
CURRENT_DIR = Path.cwd().resolve()


# شناسایی خودکار ریشه پروژه
if CURRENT_DIR.name == "notebooks":

    PROJECT_ROOT = CURRENT_DIR.parent

elif (
    (CURRENT_DIR / "notebooks").exists()
    and (CURRENT_DIR / "data").exists()
):

    PROJECT_ROOT = CURRENT_DIR

else:

    raise FileNotFoundError(
        "Project root could not be detected. "
        "Please run this notebook from the project folder or the notebooks folder."
    )


PROCESSED_FOLDER = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed"
)

TABLES_FOLDER = os.path.join(
    PROJECT_ROOT,
    "outputs",
    "tables"
)

MODELS_FOLDER = os.path.join(
    PROJECT_ROOT,
    "models"
)


print("Project root:")
print(PROJECT_ROOT)

print("\nProcessed folder:")
print(PROCESSED_FOLDER)

print("\nTables folder:")
print(TABLES_FOLDER)

print("\nModels folder:")
print(MODELS_FOLDER)


print("\nPath check:")

print(
    "Processed folder exists:",
    os.path.exists(PROCESSED_FOLDER)
)

print(
    "Tables folder exists:",
    os.path.exists(TABLES_FOLDER)
)

print(
    "Models folder exists:",
    os.path.exists(MODELS_FOLDER)
)
```

```python
# در این سلول وجود تمام فایل‌های اصلی مورد نیاز برای استخراج نتایج مقاله بررسی می‌شود.

required_result_files = [

    # Construct Harmonization
    "B_EFL_Distribution_Summary.xlsx",
    "B_EFL_SMD_Summary.xlsx",
    "B_EFL_Correlation_Comparison.xlsx",
    "B_EFL_Final_Harmonization_Evidence.xlsx",

    # Model Development
    "Model_Tuning_Summary.xlsx",
    "Model_Cross_Validation_Performance.xlsx",
    "Model_Test_Performance.xlsx",

    # External Validation
    "External_Validation_Primary_Performance.xlsx",
    "Internal_vs_External_Performance.xlsx",
    "External_Validation_Sensitivity_Analysis.xlsx",

    # Extended Predictor Analysis
    "Dataset_B_Predictor_Sets_All_Algorithms.xlsx",
    "Dataset_B_RMSE_Predictor_Set_Comparison.xlsx"
]


missing_files = []


for file_name in required_result_files:

    file_path = os.path.join(
        TABLES_FOLDER,
        file_name
    )

    if os.path.exists(file_path):

        print(
            "✓ Found:",
            file_name
        )

    else:

        print(
            "✗ Missing:",
            file_name
        )

        missing_files.append(
            file_name
        )


print("\n--------------------------------")

if len(missing_files) == 0:

    print(
        "All required result files are available."
    )

else:

    print(
        "Number of missing files:",
        len(missing_files)
    )
```

```python
# در این سلول فایل‌های نهایی Quality Control برای Dataset B و Dataset EFL بارگذاری می‌شوند.

B_QUALITY_PATH = os.path.join(
    PROCESSED_FOLDER,
    "dataset_B_quality_flagged.csv"
)

EFL_QUALITY_PATH = os.path.join(
    PROCESSED_FOLDER,
    "dataset_EFL_quality_flagged.csv"
)


df_B_quality = pd.read_csv(
    B_QUALITY_PATH
)

df_EFL_quality = pd.read_csv(
    EFL_QUALITY_PATH
)


print(
    "Dataset B shape:",
    df_B_quality.shape
)

print(
    "Dataset EFL shape:",
    df_EFL_quality.shape
)


print("\nDataset B quality flags:")

for column in [
    "Straight_Line_Flag",
    "Very_Fast_Flag",
    "High_Risk_Flag",
    "Strict_Duplicate_Flag"
]:

    print(
        column,
        "=",
        int(df_B_quality[column].sum())
    )


print("\nDataset EFL quality flags:")

for column in [
    "Straight_Line_Flag",
    "Exact_Duplicate_Flag"
]:

    print(
        column,
        "=",
        int(df_EFL_quality[column].sum())
    )
```

```python
# در این سلول خلاصه نهایی حجم نمونه و Quality Flags برای Dataset B و Dataset EFL ساخته می‌شود.

quality_summary = pd.DataFrame({

    "Dataset": [
        "Dataset B",
        "Dataset EFL"
    ],

    "Role": [
        "Model Development",
        "External Validation"
    ],

    "Primary_N": [
        len(df_B_quality),
        len(df_EFL_quality)
    ],

    "Straight_Line_Flag_N": [
        int(
            df_B_quality[
                "Straight_Line_Flag"
            ].sum()
        ),
        int(
            df_EFL_quality[
                "Straight_Line_Flag"
            ].sum()
        )
    ],

    "Duplicate_Flag_N": [
        int(
            df_B_quality[
                "Strict_Duplicate_Flag"
            ].sum()
        ),
        int(
            df_EFL_quality[
                "Exact_Duplicate_Flag"
            ].sum()
        )
    ],

    "Primary_Exclusion_Based_on_Flags": [
        0,
        0
    ]
})


quality_summary[
    "Straight_Line_Percent"
] = (
    quality_summary[
        "Straight_Line_Flag_N"
    ]
    / quality_summary[
        "Primary_N"
    ]
    * 100
).round(2)


quality_summary[
    "Duplicate_Flag_Percent"
] = (
    quality_summary[
        "Duplicate_Flag_N"
    ]
    / quality_summary[
        "Primary_N"
    ]
    * 100
).round(2)


display(
    quality_summary
)
```

```python
# در این سلول فایل Excel مربوط به Dataset C پیدا و میزان تکرار کامل رکوردها بررسی می‌شود.

EXCLUDED_FOLDER = os.path.join(
    PROJECT_ROOT,
    "data",
    "excluded_datasets"
)

dataset_c_excel_files = [
    file_name
    for file_name in os.listdir(EXCLUDED_FOLDER)
    if file_name.lower().endswith(".xlsx")
]

print("Excel files found:")
print(dataset_c_excel_files)


if len(dataset_c_excel_files) != 1:
    raise ValueError(
        "Expected exactly one Excel file for Dataset C."
    )


DATASET_C_PATH = os.path.join(
    EXCLUDED_FOLDER,
    dataset_c_excel_files[0]
)


df_C = pd.read_excel(
    DATASET_C_PATH
)


raw_n_C = len(df_C)

unique_n_C = len(
    df_C.drop_duplicates()
)

extra_duplicate_n_C = int(
    df_C.duplicated().sum()
)

duplicate_group_rows_C = int(
    df_C.duplicated(
        keep=False
    ).sum()
)


extra_duplicate_percent_C = (
    extra_duplicate_n_C
    / raw_n_C
    * 100
)


print("\nDataset C shape:")
print(df_C.shape)

print("\nRaw rows:")
print(raw_n_C)

print("Unique full rows:")
print(unique_n_C)

print("Extra duplicated rows:")
print(extra_duplicate_n_C)

print(
    "Rows belonging to duplicate groups:",
    duplicate_group_rows_C
)

print(
    "Extra duplicate percentage:",
    round(extra_duplicate_percent_C, 2),
    "%"
)
```

```python
# در این سلول نقش نهایی سه Dataset و تصمیم استفاده یا عدم استفاده از آن‌ها در مطالعه ثبت می‌شود.

dataset_status_summary = pd.DataFrame({

    "Dataset": [
        "Dataset B",
        "Dataset EFL",
        "Dataset C"
    ],

    "Raw_N": [
        len(df_B_quality),
        len(df_EFL_quality),
        raw_n_C
    ],

    "Study_Role": [
        "Model Development",
        "External Validation",
        "Excluded Candidate Dataset"
    ],

    "Included_in_Primary_Analysis": [
        "Yes",
        "Yes",
        "No"
    ],

    "Reason_or_QC_Decision": [
        "Quality flags retained; evaluated through sensitivity-oriented QC",
        "Quality flags retained; external sensitivity analysis performed",
        (
            f"Excluded because only {unique_n_C} of {raw_n_C} full rows "
            f"were unique; {extra_duplicate_n_C} extra duplicated rows "
            f"({extra_duplicate_percent_C:.2f}%) were detected"
        )
    ]
})


display(
    dataset_status_summary
)
```

```python
# در این سلول جداول نهایی مربوط به Construct Harmonization بین Dataset B و Dataset EFL بارگذاری می‌شوند.

distribution_summary = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "B_EFL_Distribution_Summary.xlsx"
    )
)

smd_summary = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "B_EFL_SMD_Summary.xlsx"
    )
)

correlation_summary = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "B_EFL_Correlation_Comparison.xlsx"
    )
)

harmonization_evidence = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "B_EFL_Final_Harmonization_Evidence.xlsx"
    )
)


print("Distribution Summary:")
display(distribution_summary)

print("\nSMD Summary:")
display(smd_summary)

print("\nCorrelation Comparison:")
display(correlation_summary)

print("\nFinal Harmonization Evidence:")
display(harmonization_evidence)
```

```python
# در این سلول دو جدول خلاصه و مقاله‌پسند برای شواهد Construct Harmonization ساخته می‌شوند.

# استخراج SD هر سازه در Dataset B
sd_B = (
    distribution_summary[
        distribution_summary["Dataset"] == "B"
    ]
    .set_index("Construct")["Std"]
)

# استخراج SD هر سازه در Dataset EFL
sd_EFL = (
    distribution_summary[
        distribution_summary["Dataset"] == "EFL"
    ]
    .set_index("Construct")["Std"]
)


harmonization_article_table = pd.DataFrame({

    "Construct": [
        "CON",
        "SAT",
        "CI"
    ],

    "Role": [
        "Predictor",
        "Predictor",
        "Target"
    ],

    "Mean_B": [
        3.822,
        3.940,
        3.959
    ],

    "SD_B": [
        sd_B["CON_score"],
        sd_B["SAT_score"],
        sd_B["CI_score"]
    ],

    "Mean_EFL": [
        3.804,
        3.856,
        3.887
    ],

    "SD_EFL": [
        sd_EFL["CON_score"],
        sd_EFL["SAT_score"],
        sd_EFL["CI_score"]
    ],

    "Absolute_SMD": [
        0.027,
        0.125,
        0.108
    ],

    "Semantic_Compatibility": [
        "Good",
        "Very High",
        "Very High"
    ],

    "Cross_Dataset_Use": [
        "Yes",
        "Yes",
        "Yes"
    ]
})


harmonization_article_table[
    [
        "Mean_B",
        "SD_B",
        "Mean_EFL",
        "SD_EFL",
        "Absolute_SMD"
    ]
] = harmonization_article_table[
    [
        "Mean_B",
        "SD_B",
        "Mean_EFL",
        "SD_EFL",
        "Absolute_SMD"
    ]
].round(3)


correlation_article_table = correlation_summary.copy()

correlation_article_table = correlation_article_table[
    [
        "Relationship",
        "r_B",
        "r_EFL",
        "Absolute_Difference",
        "Fisher_Z",
        "P_Value"
    ]
].round(4)


print("Article Harmonization Summary:")
display(
    harmonization_article_table
)

print("\nCross-Dataset Correlation Stability:")
display(
    correlation_article_table
)
```

```python
# در این سلول جداول نهایی مربوط به تنظیم، Cross-Validation و ارزیابی داخلی مدل‌ها از Notebook 3 بارگذاری می‌شوند.

model_tuning_summary = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "Model_Tuning_Summary.xlsx"
    )
)

model_cv_summary = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "Model_Cross_Validation_Performance.xlsx"
    )
)

model_test_summary = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "Model_Test_Performance.xlsx"
    )
)


print("Model Tuning Summary:")
display(
    model_tuning_summary
)


print("\nCross-Validation Performance:")
display(
    model_cv_summary
)


print("\nInternal Test Performance:")
display(
    model_test_summary
)
```

```python
# در این سلول نتایج Cross-Validation و Test Set پنج مدل در یک جدول نهایی برای مقاله ادغام می‌شوند.

model_test_without_baseline = model_test_summary[
    model_test_summary["Model"] != "Baseline"
].copy()


article_model_comparison = model_cv_summary.merge(
    model_test_without_baseline,
    on="Model",
    how="left"
)


article_model_comparison = article_model_comparison.rename(
    columns={
        "MAE": "Test_MAE",
        "RMSE": "Test_RMSE",
        "R2": "Test_R2"
    }
)


article_model_comparison[
    "CV_RMSE_Rank"
] = (
    article_model_comparison[
        "CV_RMSE_Mean"
    ]
    .rank(
        method="min",
        ascending=True
    )
    .astype(int)
)


article_model_comparison[
    "Selected_Final_Model"
] = np.where(
    article_model_comparison["Model"] == "Linear Regression",
    "Yes",
    "No"
)


article_model_comparison = article_model_comparison[
    [
        "Model",
        "CV_MAE_Mean",
        "CV_MAE_Std",
        "CV_RMSE_Mean",
        "CV_RMSE_Std",
        "CV_R2_Mean",
        "CV_R2_Std",
        "CV_RMSE_Rank",
        "Test_MAE",
        "Test_RMSE",
        "Test_R2",
        "Selected_Final_Model"
    ]
]


display(
    article_model_comparison
)
```

```python
# در این سلول عملکرد Baseline به‌صورت جداگانه برای مقایسه با مدل نهایی استخراج می‌شود.

baseline_summary = model_test_summary[
    model_test_summary["Model"] == "Baseline"
].copy()


baseline_summary = baseline_summary.rename(
    columns={
        "MAE": "Test_MAE",
        "RMSE": "Test_RMSE",
        "R2": "Test_R2"
    }
)


display(
    baseline_summary
)
```

```python
# در این سلول نتایج اصلی External Validation، مقایسه داخلی-خارجی و Sensitivity Analysis بارگذاری می‌شوند.

external_primary = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "External_Validation_Primary_Performance.xlsx"
    )
)

internal_external = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "Internal_vs_External_Performance.xlsx"
    )
)

external_sensitivity = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "External_Validation_Sensitivity_Analysis.xlsx"
    )
)


print("Primary External Validation:")
display(
    external_primary
)

print("\nInternal vs External Performance:")
display(
    internal_external
)

print("\nExternal Validation Sensitivity Analysis:")
display(
    external_sensitivity
)
```

```python
# در این سلول عملکرد مدل نهایی در Test Set داخلی و Dataset خارجی در یک جدول مقاله‌پسند آماده می‌شود.

article_external_validation = internal_external.copy()


numeric_columns = [
    column
    for column in [
        "MAE",
        "RMSE",
        "R2"
    ]
    if column in article_external_validation.columns
]


article_external_validation[
    numeric_columns
] = article_external_validation[
    numeric_columns
].round(4)


display(
    article_external_validation
)
```

```python
# در این سلول نتایج Sensitivity Analysis اعتبارسنجی خارجی برای گزارش نهایی مقاله مرتب می‌شوند.

article_sensitivity = external_sensitivity[
    [
        "Scenario",
        "N",
        "MAE",
        "RMSE",
        "R2"
    ]
].copy()


article_sensitivity[
    [
        "MAE",
        "RMSE",
        "R2"
    ]
] = article_sensitivity[
    [
        "MAE",
        "RMSE",
        "R2"
    ]
].round(4)


display(
    article_sensitivity
)
```

```python
# در این سلول نتایج نهایی بررسی Predictorهای اضافی Dataset B بارگذاری می‌شوند.

extended_all_models = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "Dataset_B_Predictor_Sets_All_Algorithms.xlsx"
    )
)

extended_rmse = pd.read_excel(
    os.path.join(
        TABLES_FOLDER,
        "Dataset_B_RMSE_Predictor_Set_Comparison.xlsx"
    ),
    index_col=0
)


print("Extended Predictor Analysis - All Models:")
display(
    extended_all_models
)

print("\nRMSE Comparison:")
display(
    extended_rmse
)
```

```python
# در این سلول بهترین Predictor Set هر الگوریتم و میزان بهبود آن نسبت به مدل Core خلاصه می‌شود.

extended_summary_rows = []


for algorithm in extended_rmse.index:

    row = extended_rmse.loc[
        algorithm
    ]

    core_rmse = row[
        "M1_Core"
    ]

    best_predictor_set = row.idxmin()

    best_rmse = row.min()

    improvement_vs_core = (
        core_rmse - best_rmse
    )


    if best_predictor_set == "M1_Core":

        interpretation = (
            "Core model remained best"
        )

    elif improvement_vs_core < 0.005:

        interpretation = (
            "Only marginal improvement over Core"
        )

    else:

        interpretation = (
            "Improvement over Core"
        )


    extended_summary_rows.append({

        "Algorithm":
            algorithm,

        "Best_Predictor_Set":
            best_predictor_set,

        "Core_RMSE":
            core_rmse,

        "Best_RMSE":
            best_rmse,

        "RMSE_Improvement_vs_Core":
            improvement_vs_core,

        "Interpretation":
            interpretation
    })


article_extended_summary = pd.DataFrame(
    extended_summary_rows
)


article_extended_summary[
    [
        "Core_RMSE",
        "Best_RMSE",
        "RMSE_Improvement_vs_Core"
    ]
] = article_extended_summary[
    [
        "Core_RMSE",
        "Best_RMSE",
        "RMSE_Improvement_vs_Core"
    ]
].round(4)


display(
    article_extended_summary
)
```

```python
# در این سلول تمام جداول نهایی پروژه در یک فایل Excel مرجع برای نگارش مقاله ذخیره می‌شوند.

MASTER_RESULTS_PATH = os.path.join(
    TABLES_FOLDER,
    "Article_Master_Results.xlsx"
)


with pd.ExcelWriter(
    MASTER_RESULTS_PATH,
    engine="openpyxl"
) as writer:

    quality_summary.to_excel(
        writer,
        sheet_name="Quality_Control",
        index=False
    )

    dataset_status_summary.to_excel(
        writer,
        sheet_name="Dataset_Status",
        index=False
    )

    harmonization_article_table.to_excel(
        writer,
        sheet_name="Harmonization",
        index=False
    )

    correlation_article_table.to_excel(
        writer,
        sheet_name="Correlations",
        index=False
    )

    model_tuning_summary.to_excel(
        writer,
        sheet_name="Model_Tuning",
        index=False
    )

    article_model_comparison.to_excel(
        writer,
        sheet_name="Model_Comparison",
        index=False
    )

    baseline_summary.to_excel(
        writer,
        sheet_name="Baseline",
        index=False
    )

    article_external_validation.to_excel(
        writer,
        sheet_name="External_Validation",
        index=False
    )

    article_sensitivity.to_excel(
        writer,
        sheet_name="Sensitivity",
        index=False
    )

    article_extended_summary.to_excel(
        writer,
        sheet_name="Extended_Predictors",
        index=False
    )

    extended_rmse.to_excel(
        writer,
        sheet_name="Extended_RMSE"
    )


print("Article Master Results saved successfully:")
print(MASTER_RESULTS_PATH)
```

```python
# در این سلول ایجاد صحیح فایل نهایی Master Results و Sheetهای آن بررسی می‌شود.

if os.path.exists(
    MASTER_RESULTS_PATH
):

    print(
        "✓ Article_Master_Results.xlsx exists."
    )

    master_excel = pd.ExcelFile(
        MASTER_RESULTS_PATH
    )

    print("\nSheets:")

    for sheet_name in master_excel.sheet_names:
        print(
            "✓",
            sheet_name
        )

else:

    print(
        "✗ Article_Master_Results.xlsx was not created."
    )
```
